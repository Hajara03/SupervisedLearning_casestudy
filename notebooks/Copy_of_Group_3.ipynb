```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

1.Loading dataset

```python
df=pd.read_csv('/content/drive/MyDrive/ICT_AI ML/House_Pricing.csv')
df.head()
```

2.Data Visualisation

```python
# Visualisation of Sale Price distribution using Histplot
plt.figure(figsize=(10, 8))
sns.histplot(df['Sale Price'].dropna(),kde=True,bins=30,color='red')
plt.xlabel('Sale Price')
plt.ylabel('count')
plt.show()
```

```python
# Visualisation of Sale Price vs Flat Area using Scatterplot
plt.figure(figsize=(10,8))
sns.scatterplot(data=df,x='Flat Area (in Sqft)',y='Sale Price',alpha=0.9, color='red')
plt.xlabel('Flat Area (in Sqft)')
plt.ylabel('Sale Price')
plt.show()
```

```python
#Visualisation Number of Bedrooms vs Sale Price using Barplot
plt.figure(figsize=(10,5))
sns.barplot(data=df,x='No of Bedrooms',y='Sale Price',errorbar=None)
plt.xlabel('Number of Bedrooms')
plt.ylabel('Sale Price')
plt.show()
```
